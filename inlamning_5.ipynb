## Introduction

Earth's surface temperature is set by its energy balance: the incoming solar radiation absorbed by the planet must, on average, equal the outgoing radiation it emits back to space. One important part of this balance is Earth's **albedo** — the fraction of incoming sunlight reflected back to space — and how strongly it depends on the extent of ice and snow. Since temperature controls how much ice and snow exist, there is a **positive feedback loop** between surface temperature and albedo: a colder planet has more ice, which reflects more sunlight, which cools the planet further (and vice versa for a warmer planet).

This feedback can create **tipping points**: thresholds where the system shifts abruptly and, on human timescales, irreversibly, between two extreme states. The two extreme states are a planet with no polar ice at all, and a planet that is entirely ice-covered ("Snowball Earth").

### Task B: Hysteresis

A system shows **hysteresis** when its current state depends not only on the external conditions acting on it right now, but also on the *path* it took to get there — on its history. Hysteresis is closely linked to **bistability**: if a system has two different stable equilibria for the exact same external conditions, then which one it ends up in depends on where it started or on which direction a parameter was changed from. Parts of the real climate system (such as ocean circulation patterns or ice sheets) can show this behaviour: pushing a system past a tipping point in one direction does not guarantee that reversing the push by the same amount will undo the change. Note that hysteresis is a property of the *equilibrium* behaviour of a system, not of how you compute it — if you ever find different answers depending on how you searched for the equilibrium, that is a symptom of hysteresis in the underlying system, not a bug in your method. Not every system with a sharp, tipping-point-like transition necessarily shows hysteresis, and not every system that changes gradually is free of it — the two properties are related but not identical.
